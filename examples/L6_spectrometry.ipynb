```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv("../data_sets/spectrometer_dataset_2.csv")
# print(df.head(10))
data = df.to_numpy()
x = data[:, 0:8]
y = data[:, 8]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.33)

# We need to mean-center the data. StandardScalar will mean-center the data
# It also normalizes the variance per feature (column)
scalar = StandardScaler()
scalar.fit(x_train)
u_train = scalar.transform(x_train)  # Standard-normalized dataset (training only)
```

```python
C = PCA(2)
C.fit(u_train)
# print(C.components_)
# print(C.explained_variance_)
x_train_pca = C.transform(u_train)
# print(x_train_pca.shape, u_train.shape)
```

```python
# Model training
model = LogisticRegression()
model.fit(x_train_pca, y_train)

# Test the model
u_test = scalar.transform(x_test)  # Standard normalization applied on test set
x_test_pca = C.transform(u_test)  # Apply PCA on test set after standard normalization
y_pred = model.predict(x_test_pca)
print(confusion_matrix(y_test, y_pred), accuracy_score(y_test, y_pred))
```

```python
# Testing without dimension reduction
model2 = LogisticRegression()
model2.fit(x_train, y_train)
y_pred2 = model2.predict(x_test)
print(confusion_matrix(y_test, y_pred2), accuracy_score(y_test, y_pred2))
```
