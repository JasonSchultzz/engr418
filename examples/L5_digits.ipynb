```python
# IMPORTANT FOR QUIZ 2!!!!!!!!

import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score
from PIL import Image

folder_training = ["../MNIST/training/" + str(i) + "/" for i in range(10)]  # Create a list of paths
folder_testing = ["../MNIST/testing/" + str(i) + "/" for i in range(10)]

def get_data(folder: str, im_width, label, n_samples):
    file_names = os.listdir(folder)
    x = np.empty((n_samples, im_width**2))
    y = np.empty((n_samples, 1))

    for i in range(n_samples):
        path = folder + file_names[i]
        im = Image.open(path).convert("L")  # convert to greyscale
        im = im.resize((im_width, im_width))
        im_array = np.asarray(im)
        x[i:] = im_array.reshape(1, -1)
        y[i, 0] = label
    return x, y
```

```python
im_width = 10

P_c = 1000
x_train = np.empty((P_c*10, im_width**2))
y_train = np.empty((P_c*10, 1))
for i in range(10):
    x_i, y_i = get_data(folder_training[i], im_width, i, P_c)
    x_train[i*P_c:(i+1)*P_c, :] = x_i
    y_train[i*P_c:(i+1)*P_c, :] = y_i

print(x_train.shape, y_train.shape)
```

```python
# Choosing to show an image of a number by changing the row number.
# Each set of 1000 will contain one digit.
plt.imshow(x_train[1000,:].reshape(im_width, im_width), cmap = "gray")
```

```python
model = LogisticRegression()
model.fit(x_train, y_train)

print(model.coef_.shape)
plt.imshow(model.coef_[0,:].reshape(im_width, im_width), cmap = "gray")

P_c = 100  # Load 100 test images per digit
x_test = np.empty((P_c*10, im_width**2))
y_test = np.empty((P_c*10, 1))
for i in range(10):
    x_i, y_i = get_data(folder_testing[i], im_width, i, P_c)
    x_test[i*P_c:(i+1)*P_c, :] = x_i
    y_test[i*P_c:(i+1)*P_c, :] = y_i

y_pred = model.predict(x_test)
print(confusion_matrix(y_test, y_pred))
print(accuracy_score(y_test, y_pred))
```
