```python
# importing numpy with alias np
import numpy as np
# create a random 2x4 array with uniform distribution on [0,1)
a = np.random.random((2,4))
# print the array
print('array:\n', a)
print('array at (0,0) and (1,3):\n', a[(0,0)], a[(1,3)])
```

```python
import pandas as pd
import numpy as np
# data frame with 10 rows and 4 cols, random normal elments
# column labels A, B, C, and D
df = pd.DataFrame(np.random.randn(10,4), columns=list('ABCD'))
print(df)
```

```python
import matplotlib.pyplot as plt
# plot column A vs the array 0,1,2,...,9
plt.plot(list(range(0,10)),df['A'])
plt.title('Data')
plt.xlabel('index')
plt.ylabel('value')
```
