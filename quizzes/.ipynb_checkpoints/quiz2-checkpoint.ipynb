```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('transmission.csv')
data = df.to_numpy()

#-----------------------------------
# write your code for question (1) here:
x_train = data[0:600, 0:8]
y_train = data[0:600, 8]

K = 8
scalar = StandardScaler()
scalar.fit(x_train)
u_train = scalar.transform(x_train)
C = PCA(K)
C.fit(u_train)

#-----------------------------------
# write your code for question (2) here:
x_train_pca = C.transform(u_train)

x_test = data[600:, 0:8]
y_test = data[600:, 8]
u_test = scalar.transform(x_test)
x_test_pca = C.transform(u_test)

model = LogisticRegression()
model.fit(x_train_pca, y_train)
y_pred = model.predict(x_test_pca)

print(accuracy_score(y_test, y_pred))

#-----------------------------------
# write your accuracy and answer question (3) here:
'''
K    Accuracy Score
1    0.6067
2    0.8233
3    0.9767
4    0.9800
5    0.9767
6    0.9800
7    0.9800
8    0.9833

Therefore, performance of the model seems to cap at around K = 3, so that is the best choice
in terms of both accuracy and computational performance
'''
```
